## Задание 1: Зеркальное отражение цифр

Пользователь вводит пятизначное число. Программа должна зеркально отразить три центральные цифры. Первая и последняя остаются на местах.

<br>

**Примеры:**
* `23456 -> 25436`
* `30789 -> 38709`

```python
LENGTH = 5
while True:
    digits = input(f"Введите {LENGTH}-значное число: ").strip()
    if digits.isdigit() and len(digits) == LENGTH:
        break
    print(f"Нужно ровно {LENGTH} цифр.")

print("Результат:", digits[0] + digits[1:-1][::-1] + digits[-1])
```

## Задание 2: Подсчет выходных до отпуска

Пользователь вводит сколько дней осталось до ближайшего отпуска. Программа должна вывести количество выходных дней до отпуска, если учесть, что выходные это суббота и воскресенье, сегодня понедельник и праздники мы не учитываем.  

<br>

**Примеры:**
* `4  -> 0`
* `6  -> 1`
* `14 -> 4`

```python
# ВАШ КОД ЗДЕСЬ
days = int(input("Сколько дней осталось до отпуска: "))
weekends = (days // 7) * 2 + min(max(days % 7 - 4, 0), 2)
print("Выходных дней до отпуска:", weekends)
```

## Задание 3: Плитка шоколада

Пользователь вводит длину и ширину плитки шоколада, а также размер куска, который хочет отломить. Программа должна вычислить - можно ли совершить подобный разлом или нет, если учесть, что ломать можно только по прямой.

<br>

**Примеры:**
* `3, 4, 6  -> True`
* `5, 7, 8  -> False`
* `4, 5, 12 -> True`

```python
# ВАШ КОД ЗДЕСЬ
length, width, piece = (int(input(f"{p}: ")) for p in ("Длина", "Ширина", "Кусок"))
result = 0 < piece <length * width and (piece % length == 0 or piece % width == 0)
print(result)
```

## Задание 4: Римские числа

Пользователь вводит целое положительное число, программа должна вернуть строку в виде римского числа.

<br>

**Примеры:**
* `3   -> III`
* `15  -> XV`
* `234 -> CCXXXIV`

```python
# ВАШ КОД ЗДЕСЬ
values = [
        (1000, 'M'), (900, 'CM'), (500, 'D'), (400, 'CD'),
        (100, 'C'), (90, 'XC'), (50, 'L'), (40, 'XL'),
        (10, 'X'), (9, 'IX'), (5, 'V'), (4, 'IV'), (1, 'I')
    ]
num = int(input("Введите целое положительное число: "))
result = ''
for value, symbol in values:
    while num >= value:
        result += symbol
        num -= value
print(result)
```

## Задание 5: Проверка на вещественное число

Пользователь вводит данные, проверить - являются ли они положительным вещественным числом. Не использовать встроенные функции для проверки, только методы данных и конструкцию IF. (Дополнительное задание, по желанию - проверка на отрицательные вещественные числа)

<br>

**Примеры:**
* `5.6  -> True`
* `.78  -> True`
* `.67. -> False`
* `5    -> True`

```python
# ВАШ КОД ЗДЕСЬ
data = input("Введите данные: ").strip()

digits = "0123456789"
is_valid = True

if len(data) == 0:
    is_valid = False

sign = "+"
if is_valid and data[0] == "-":
    sign = "-"
    data = data[1:]
elif is_valid and data[0] == "+":
    data = data[1:]

if len(data) == 0:
    is_valid = False

parts = data.split(".")
if len(parts) > 2:
    is_valid = False

if is_valid:
    for part in parts:
        if len(part) == 0:
            is_valid = False
            break
        for ch in part:
            if ch not in digits:
                is_valid = False
                break
        if not is_valid:
            break

if is_valid:
    if sign == "-":
        print("True - Отрицательное вещественное число")
    else:
        print("True - Положительное вещественное число")
else:
    print("False - Это не вещественное число")
```
