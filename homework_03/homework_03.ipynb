## Задание 1: Конвертер регистров

Написать функцию, которая будет переводить snake_case в PascalCase и наоборот. 

Функция должна сама определять - какой формат ей передали. Можно добавить ключевой аргумент, который будет принудительно возвращать один из форматов.

<br>

**Примеры:**
* `otus_course     -> OtusCourse`
* `PythonIsTheBest -> python_is_the_best`



```python
# ВАШ КОД ЗДЕСЬ
def convert(s):
    is_snake = "_" in s and s == s.lower()
    is_pascal = "_" not in s and s[:1].isupper()

    if not is_snake and not is_pascal:
        return None

    if is_snake:
        return "".join(word.capitalize() for word in s.split("_"))
    else:
        result = []
        for i, ch in enumerate(s):
            if ch.isupper() and i > 0:
                result.append("_")
            result.append(ch.lower())
        return "".join(result)

for value in ["otus_course", "PythonIsTheBest", "otus course", "Python_Is_The_Best"]:
    res = convert(value)
    if res is None:
        print(f"Не удалось определить формат: {value!r}")
    else:
        print(f"{value} -> {res!r}")
```

## Задание 2: Проверка валидности даты

Написать функцию проверяющую валидность введенной даты.

<br>

**Примеры:**
* `29.02.2000 -> True`
* `29.02.2001 -> False`
* `31.04.1962 -> False`



```python
# ВАШ КОД ЗДЕСЬ
def is_valid_date(day, month, year):
    if month < 1 or month > 12:
        return False
    if year < 1:
        return False

    days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        days_in_month[1] = 29

    if day < 1 or day > days_in_month[month - 1]:
        return False
    return True

date_inpute = input("Введите дату в формате: ДД.ММ.ГГГГ")

parts = date_inpute.split(".")
if len(parts) != 3:
    print ("Неверный формат!!!")
else:
    if parts[0].isdigit() and parts[1].isdigit() and parts[2].isdigit():
        day = int(parts[0])
        month = int(parts[1])
        year = int(parts[2])

        print(is_valid_date(day, month, year))

    else:
        print("Неверный формат!!!")
```

## Задание 3: Проверка на простое число

Функция проверки на простое число. Простые числа – это такие числа, которые делятся на себя и на единицу.

<br>

**Примеры:**
* `17 -> True`
* `20 -> False`
* `23 -> True`

```python
# ВАШ КОД ЗДЕСЬ
def is_prime(num):
    if num < 2:
        return False
    if num == 2:
        return True
    if num % 2 == 0:
        return False
    d = 3
    while d * d <= num:
        if num % d == 0:
            return False
        d += 2
    return True

for value in [0, 1, 2, 5, 9, 17, 20, 23, 97, 100]:
    print(f"{value:4} -- {is_prime(value)!r}")
```

## Задание 4: Учет пользователей

Пользователь в бесконечном цикле вводит данные пользователей: имя, затем фамилию, возраст и ID. Ввод продолжается до тех пор, пока не будет введено пустое поле. 

Пользователи заносятся в словарь, где ключ это ID пользователя, а остальные данные записываются в виде кортежа. 

**Программа должна проверять:**
* имя и фамилия состоят только из символов и начинаются с большой буквы - если не с большой, то заменяет букву на большую;
* возраст должен быть числом от 18 до 60;
* ID - целое число, дополненное до 8 знаков незначащими нулями, ID должен быть уникальным.

**Дополнительно:** написать функцию, которая будет выводить полученный словарь в виде таблицы.

```python
# ВАШ КОД ЗДЕСЬ
def normalize_name(name):
    if not name or not name.isalpha():
        return None
    return name[0].upper() + name[1:]

def normalize_age(age_str):
    if not age_str.isdigit():
        return None
    age = int(age_str)
    if 18 <= age <= 60:
        return age
    return None

def normalize_id(id_str):
    if not id_str.isdigit():
        return None
    return id_str.zfill(8)

def input_user(users):
    name = input("Имя: ").strip()
    if name == "":
        return False

    surname = input("Фамилия: ").strip()
    if surname == "":
        return False

    age_str = input("Возраст: ").strip()
    if age_str == "":
        return False

    id_str = input("ID: ").strip()
    if id_str == "":
        return False

    name_n = normalize_name(name)
    if name_n is None:
        print("Ошибка: Имя должно состоять только из букв")
        return True

    surname_n = normalize_name(surname)
    if surname_n is None:
        print("Ошибка: Фамилия должна состоять только из букв")
        return True

    age_n = normalize_age(age_str)
    if age_n is None:
        print("Ошибка: возраст должен быть числом от 18 до 60")
        return True

    id_n = normalize_id(id_str)
    if id_n is None:
        print("Ошибка: ID должен быть целым числом")
        return True

    if id_n in users:
        print(f"Ошибка: ID {id_n} уже занят")
        return True

    users[id_n] = (name_n, surname_n, age_n)
    print(f"Пользователь добавлен: {name_n} {surname_n}, возраст {age_n}, ID {id_n}")
    return True

def print_users(users):
    if not users:
        print("Список пользователей пуст")
        return
    print(f"{'ID':<10}{'Имя':<15}{'Фамилия':<15}{'Возраст':<8}")
    print("-" * 48)

    for uid, (name, surname, age) in users.items():
        print(f"{uid:<10}{name:<15}{surname:<15}{age:<8}")


users = {}

print("Введите данные пользователей (пустое поле - конец ввода):")
while input_user(users):
    pass

print()
print_users(users)
```
